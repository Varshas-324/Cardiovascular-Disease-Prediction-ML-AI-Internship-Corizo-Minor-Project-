##**Abstract**

Cardiovascular disease is one of the major health problems worldwide, and early identification of individuals at increased risk can support timely medical attention. This project develops a machine-learning-based cardiovascular disease risk screening system using demographic, physical, clinical and lifestyle-related attributes. The study uses a large cardiovascular disease dataset containing 70,000 records and 13 original attributes.

The proposed methodology performs data-quality checking, removal of unrealistic measurements, feature engineering and exploratory data analysis before model development. Additional features such as age in years, Body Mass Index (BMI), pulse pressure, Mean Arterial Pressure (MAP) and blood-pressure ratio are derived from the original variables. Five supervised machine learning algorithms—Logistic Regression, Support Vector Machine, K-Nearest Neighbor, Decision Tree and Random Forest—are trained and compared using accuracy, precision, recall, F1-score and ROC-AUC.

The best-performing model is then evaluated on an untouched test set and on a separate 20-record testing file. Feature importance and what-if analysis are included to make the system more interpretable. Finally, the trained model is deployed through a simple interactive Gradio interface. The resulting system is intended as an educational risk-screening prototype and not as a replacement for professional medical diagnosis.

##**Feature engineering**

This is one of the parts that makes your project stronger.

From the original variables, create:


### Age in years

$$
AgeYears = \frac{AgeDays}{365.25}
$$

### BMI

$$
BMI = \frac{Weight}{Height^2}
$$

where height is converted from cm to metres.

### Pulse pressure

$$
PP = AP_{HI} - AP_{LO}
$$

### Mean Arterial Pressure

$$
MAP = \frac{AP_{HI} + 2(AP_{LO})}{3}
$$

### Blood-pressure ratio

$$
BPRatio = \frac{AP_{HI}}{AP_{LO}}
$$

These features provide additional representations of the original cardiovascular measurements.

Step 7: Preprocessing and feature engineering

```python
data = df.copy()

# Remove identifier
data = data.drop(columns=["id"])

# Age conversion
data["age_years"] = data["age"] / 365.25

# BMI
data["bmi"] = data["weight"] / ((data["height"] / 100) ** 2)

# Pulse pressure
data["pulse_pressure"] = data["ap_hi"] - data["ap_lo"]

# Mean arterial pressure
data["map"] = (data["ap_hi"] + 2 * data["ap_lo"]) / 3

# Blood pressure ratio
data["bp_ratio"] = data["ap_hi"] / data["ap_lo"]

# Remove original age measured in days
data = data.drop(columns=["age"])

print("Feature engineering completed.")
display(data.head())
```

Step 8: Data-quality filtering

```python
before = len(data)

data = data[
    (data["height"].between(120, 210)) &
    (data["weight"].between(30, 150)) &
    (data["ap_hi"].between(80, 240)) &
    (data["ap_lo"].between(40, 140)) &
    (data["ap_hi"] >= data["ap_lo"])
].copy()

after = len(data)

print("Rows before cleaning:", before)
print("Rows after cleaning:", after)
print("Rows removed:", before - after)
```

Step 9: Debug cleaned data

```python
print("Missing values:")
print(data.isnull().sum())

print("\nInfinite values:")
print(np.isinf(
    data.select_dtypes(include=np.number)
).sum())

print("\nDuplicate rows:", data.duplicated().sum())

print("\nBMI range:")
print(data["bmi"].min(), "to", data["bmi"].max())

print("\nBlood pressure range:")
print(data["ap_hi"].min(), "to", data["ap_hi"].max())

assert data.isnull().sum().sum() == 0
assert np.isfinite(
    data.select_dtypes(include=np.number)
).all().all()

assert (data["ap_hi"] >= data["ap_lo"]).all()

print("\nDATA CLEANING VALIDATION PASSED.")
```

### Preprocessing pipeline

```python
categorical_features = [
    "gender",
    "cholesterol",
    "gluc",
    "smoke",
    "alco",
    "active"
]

numeric_features = [
    "height",
    "weight",
    "ap_hi",
    "ap_lo",
    "age_years",
    "bmi",
    "pulse_pressure",
    "map",
    "bp_ratio"
]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocessing pipeline created.")
```

### Five mandatory models

```python
models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1500,
        C=1
    ),

    "SVM": LinearSVC(
        C=0.1,
        dual="auto",
        max_iter=5000
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=15,
        weights="distance"
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=6,
        min_samples_leaf=20,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=12,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    )
}

print("Five required models created.")
```

### Model training

```python
trained_models = {}

for name, classifier in models.items():

    print("\nTraining:", name)

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("classifier", classifier)
    ])

    pipeline.fit(X_train, y_train)

    trained_models[name] = pipeline

    print("Training completed.")
```

### Model evaluation function

```python
def get_score(model, X_data):

    if hasattr(model, "predict_proba"):
        return model.predict_proba(X_data)[:, 1]

    return model.decision_function(X_data)


def evaluate_model(name, model, X_data, y_data):

    predictions = model.predict(X_data)
    scores = get_score(model, X_data)

    return {
        "Model": name,
        "Accuracy": accuracy_score(y_data, predictions),
        "Precision": precision_score(y_data, predictions),
        "Recall": recall_score(y_data, predictions),
        "F1 Score": f1_score(y_data, predictions),
        "ROC-AUC": roc_auc_score(y_data, scores)
    }
```

### Validation comparison

```python
validation_results = []

for name, model in trained_models.items():

    result = evaluate_model(
        name,
        model,
        X_val,
        y_val
    )

    validation_results.append(result)

validation_df = pd.DataFrame(validation_results)

validation_df = validation_df.sort_values(
    by="ROC-AUC",
    ascending=False
)

display(validation_df.round(4))
```

### Random Forest hyperparameter tuning

```python
rf_variants = []

for depth in [8, 12, 16]:

    for leaf in [2, 5, 10]:

        rf = RandomForestClassifier(
            n_estimators=300,
            max_depth=depth,
            min_samples_leaf=leaf,
            random_state=42,
            n_jobs=-1
        )

        pipeline = Pipeline([
            ("preprocessor", preprocessor),
            ("classifier", rf)
        ])

        pipeline.fit(X_train, y_train)

        result = evaluate_model(
            f"RF_depth_{depth}_leaf_{leaf}",
            pipeline,
            X_val,
            y_val
        )

        rf_variants.append(
            (result, pipeline)
        )

rf_tuning_df = pd.DataFrame(
    [x[0] for x in rf_variants]
)

display(
    rf_tuning_df.sort_values(
        "ROC-AUC",
        ascending=False
    ).head(10)
)
```

### Select tuned Random Forest

```python
best_rf_result, best_rf_model = max(
    rf_variants,
    key=lambda x: x[0]["ROC-AUC"]
)

print("Best Random Forest configuration:")
print(best_rf_result)
```

## Deployment prediction function

```python
import pandas as pd
import joblib

# Load the trained Random Forest pipeline
model = joblib.load("cardiovascular_risk_model.pkl")


def application_prediction(
    age,
    gender,
    height,
    weight,
    systolic_bp,
    diastolic_bp,
    cholesterol,
    glucose,
    smoking,
    alcohol,
    active
):

    # ==========================================
    # 1. INPUT VALIDATION
    # ==========================================

    if age is None or age < 18 or age > 100:
        return (
            "Invalid Input",
            "Age must be between 18 and 100 years.",
            "Check Age"
        )

    if height is None or height < 120 or height > 210:
        return (
            "Invalid Input",
            "Height must be between 120 and 210 cm.",
            "Check Height"
        )

    if weight is None or weight < 30 or weight > 150:
        return (
            "Invalid Input",
            "Weight must be between 30 and 150 kg.",
            "Check Weight"
        )

    if systolic_bp is None or systolic_bp < 80 or systolic_bp > 240:
        return (
            "Invalid Input",
            "Systolic blood pressure must be between 80 and 240 mmHg.",
            "Check Blood Pressure"
        )

    if diastolic_bp is None or diastolic_bp < 40 or diastolic_bp > 140:
        return (
            "Invalid Input",
            "Diastolic blood pressure must be between 40 and 140 mmHg.",
            "Check Blood Pressure"
        )

    if systolic_bp <= diastolic_bp:
        return (
            "Invalid Input",
            "Systolic blood pressure must be higher than diastolic blood pressure.",
            "Check Blood Pressure"
        )


    # ==========================================
    # 2. CONVERT USER INPUT TO DATASET ENCODING
    # ==========================================

    # Gender
    gender_value = 1 if gender == "Female" else 2

    # Cholesterol
    cholesterol_map = {
        "Normal": 1,
        "Above Normal": 2,
        "Well Above Normal": 3
    }

    cholesterol_value = cholesterol_map[cholesterol]

    # Glucose
    glucose_map = {
        "Normal": 1,
        "Above Normal": 2,
        "Well Above Normal": 3
    }

    glucose_value = glucose_map[glucose]

    # Smoking
    smoke_value = 1 if smoking == "Yes" else 0

    # Alcohol
    alco_value = 1 if alcohol == "Yes" else 0

    # Physical activity
    active_value = 1 if active == "Yes" else 0


    # ==========================================
    # 3. FEATURE ENGINEERING
    # ==========================================

    age_years = age

    bmi = weight / ((height / 100) ** 2)

    pulse_pressure = systolic_bp - diastolic_bp

    mean_arterial_pressure = (
        systolic_bp + (2 * diastolic_bp)
    ) / 3

    bp_ratio = systolic_bp / diastolic_bp


    # ==========================================
    # 4. CREATE DATAFRAME WITH EXACT
    #    TRAINING FEATURE NAMES
    # ==========================================

    input_data = pd.DataFrame([{

        # Categorical features
        "gender": gender_value,
        "cholesterol": cholesterol_value,
        "gluc": glucose_value,
        "smoke": smoke_value,
        "alco": alco_value,
        "active": active_value,

        # Numeric features
        "height": height,
        "weight": weight,
        "ap_hi": systolic_bp,
        "ap_lo": diastolic_bp,
        "age_years": age_years,
        "bmi": bmi,
        "pulse_pressure": pulse_pressure,
        "map": mean_arterial_pressure,
        "bp_ratio": bp_ratio

    }])


    # ==========================================
    # 5. CHECK FEATURE NAMES
    # ==========================================

    expected_features = [
        "gender",
        "cholesterol",
        "gluc",
        "smoke",
        "alco",
        "active",
        "height",
        "weight",
        "ap_hi",
        "ap_lo",
        "age_years",
        "bmi",
        "pulse_pressure",
        "map",
        "bp_ratio"
    ]

    input_data = input_data[expected_features]


    # ==========================================
    # 6. MODEL PREDICTION
    # ==========================================

    prediction = model.predict(input_data)[0]


    # ==========================================
    # 7. MODEL PROBABILITY
    # ==========================================

    probability = None

    if hasattr(model, "predict_proba"):

        probabilities = model.predict_proba(input_data)[0]

        classes = list(model.classes_)

        predicted_index = classes.index(prediction)

        probability = probabilities[predicted_index] * 100


    # ==========================================
    # 8. INTERPRET MODEL RESULT
    # ==========================================

    if prediction == 1:

        result = "Higher predicted cardiovascular disease risk"

    else:

        result = "Lower predicted cardiovascular disease risk"


    # ==========================================
    # 9. HEALTH WARNING
    # ==========================================

    warnings = []

    if systolic_bp >= 180 or diastolic_bp >= 120:

        warnings.append(
            "⚠️ Very high blood pressure detected. "
            "Please seek appropriate medical attention."
        )

    elif systolic_bp >= 140 or diastolic_bp >= 90:

        warnings.append(
            "⚠️ High blood pressure detected. "
            "Professional medical evaluation is recommended."
        )


    # ==========================================
    # 10. CREATE DETAILS
    # ==========================================

    details = f"Prediction: {result}\n\n"

    if probability is not None:

        details += (
            f"Model probability for predicted class: "
            f"{probability:.2f}%\n\n"
        )

    if warnings:

        details += "\n".join(warnings) + "\n\n"

    details += (
        "This is an educational machine-learning "
        "screening prototype and is not a medical diagnosis."
    )


    # ==========================================
    # 11. CREATE FLAG
    # ==========================================

    if warnings:

        flag = warnings[0]

    elif prediction == 1:

        flag = "Higher Risk Prediction"

    else:

        flag = "Lower Risk Prediction"


    # ==========================================
    # 12. RETURN RESULTS
    # ==========================================

    return result, details, flag
```

## **Limitations**

The developed system has several limitations.

The model is trained on a single supplied cardiovascular disease dataset, so its performance may not generalize to all populations or healthcare settings.

The available attributes do not represent the complete set of clinical information that may be considered in professional cardiovascular risk assessment.

The model learns statistical associations from historical data and does not establish causal relationships between individual features and cardiovascular disease.

The 20-record testing dataset is small and is therefore useful for demonstration but not sufficient for estimating general model performance.

The Gradio application is a research and educational prototype and has not undergone clinical validation or regulatory evaluation.

Predicted probabilities generated by the machine-learning model should not be interpreted as actual clinical probabilities for an individual patient.

Advanced ensemble, boosting, calibrated-probability and deep-learning approaches can be investigated and compared with the current classical models.

Explainable AI techniques such as SHAP can be incorporated to provide instance-level explanations of individual predictions.

The system can be extended into a mobile or responsive web application.

External clinical validation can be performed using independently collected datasets before considering real-world clinical use.

Future work can also investigate fairness, calibration, uncertainty estimation, threshold selection and monitoring of model performance after deployment.

## **Conclusion**

This project developed a complete machine-learning-based cardiovascular disease risk screening prototype using demographic, physical, clinical and lifestyle-related information.

The workflow included data auditing, data-quality filtering, feature engineering, exploratory data analysis, correlation analysis, comparative evaluation of Logistic Regression, SVM, KNN, Decision Tree and Random Forest, Random Forest tuning, final testing, independent 20-record testing, feature-importance analysis and what-if experimentation.

The selected Random Forest model achieved approximately 73.42% accuracy and 0.80 ROC-AUC on the untouched test dataset in the documented execution. The separate 20-record demonstration achieved approximately 70% accuracy.

The trained model was successfully integrated into an interactive Gradio interface with user-friendly categorical inputs, input validation and prediction-probability display. The project demonstrates an end-to-end machine-learning workflow from raw data to an interactive deployed prototype.

The resulting system is intended for educational and research purposes and should not be considered a substitute for professional medical diagnosis or clinical decision-making.
